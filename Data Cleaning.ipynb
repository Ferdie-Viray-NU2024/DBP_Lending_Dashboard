2. Adding New Feature

```python
def month_name(month_id):
    if month_id == 3:
        return 'MAR'
    if month_id == 6:
        return 'JUN'
    if month_id == 9:
        return 'SEPT'
    if month_id == 12:
        return 'DEC'

df2['Month Name'] = df2['MONTH_ID'].map(month_name)
df2
```

```python
def interest_rate_range(interest_rate):
    if interest_rate < 1.00:
        return 'Below 1.00%'
    if 1.00 <= interest_rate < 2.99:
        return '1.00% - 2.99%'
    if 3.00 <= interest_rate < 4.99:
        return '3.00% - 4.99%'  
    if 5.00 <= interest_rate < 7.99:
        return '5.00% - 7.99%'  
    if 8.00 <= interest_rate < 9.99:
        return '8.00% - 9.99%'  
    if 10.00 <= interest_rate < 11.99:
        return '10.00% - 11.99%'  
    if 12.00 <= interest_rate < 14.99:
        return '12.00% - 14.99%'  
    if 15.00 <= interest_rate < 19.99:
        return '15.00% - 19.99%'  
    if interest_rate > 20.00:
        return '1.00% - 2.99%'  

df2['Interest Rate Range'] = df2['Interest Rate'].map(interest_rate_range)
df2
```

3. Capitalization

```python
df2['Location'] = df2['Location'].str.upper()
df2['Account Class'] = df2['Account Class'].str.upper()
df2
```

4. Replacing Values

```python
df2['Account/Loan Type'] = df2['Account/Loan Type'].replace({"'": "", '"': ""}, regex=True)
```

```python
AccType_grouped = df2.groupby('Account/Loan Type')['Sum of Sum of Loan Portfolio'].count()
AccType_grouped
```

```python
df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    ['AGRICULTURAL LOANS -AGRA','AGRICULTURAL LOANS -AGRI','AIC - Agrarian Reform & Other Agri Loans - AI'],
    'Agricultural Loans')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    ['IBK - Interbank Term Loans Recble - KB/UKB','IBO - Interbank Term Loans Recble - Oth Banks','INB - Interbank Term Loans Recble - NBQB'],
    'Interbank Term Loans Receivable')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    'GNF - Loans to GOCC Non-Financial',
    'Loans to GOCC')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    ['GLC - Loans to LGU - Cust Liab under TR','LLG - Loans to Local Govt Unit'],
    'Loans to LGU')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    'LNG - Loans to National Govt',
    'Loans to National Government')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    ['BA - Loans to Priv Corp - Non Fin CL BA O/S','CLT - Loans to Priv Corp -Non Fin CL under TR','LPN - Loans to Private Corp. - Non Financial'],
    'Loans to Private Corp - Non Financial')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    'LPF - Loans to Private Corp. - Financial',
    'Loans to Private Corp - Financial')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    ['MEB - SME Loans - Medium Enterprise EBP','MED - SME Loans - Medium Scale Enterprise'],
    'SME Loans - Medium')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    ['SBA - SME Loans - Small Enterprise CLBA OS','SMA - SME Loans - Small Scale Enterprise'],
    'SME Loans - Small')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    'MIC - Microfinance Loans',
    'Microfinance Loans')

df2['Account/Loan Type'] = df2['Account/Loan Type'].replace(
    'IHP - Loans to Indl for Housing Purposes',
    'Loans to Individual')
    
df2
```

```python
Region_grouped = df2.groupby('Region')['BORROWER_CODE_NAME'].count()
Region_grouped
```
